```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("train.csv")
df.head()
```

```python
df.describe()
sns.histplot(data=df,x="Age",kde=True)
plt.show()
```

```python
df.dropna(thresh=7,inplace=True)
df.describe()
```

```python
df.info()
```

```python
df.info()
sns.histplot(data=df,x="Age",kde=True)
plt.show()
```

```python
def Fillna(df):
    for col in df.columns:
        if pd.api.types.is_numeric_dtype(df[col]):
            df[col]=df[col].fillna(df[col].median())
    return df
```

```python
Fillna(df)
df.info()
```

```python
df[df['Age'].isna()]
```

```python
q1=df["Age"].quantile(0.25)
q3=df["Age"].quantile(0.75)
iqr=q3-q1
df_new=df[(df.Age>q1-1.5*iqr)&(df.Age<q3+1.5*iqr)]
```

```python
df_new.info()
```

```python
sns.histplot(data=df_new,x="Age",kde=True)
plt.show()
```
