```python
import numpy as np

# Features
X = np.array([
    [2, 2],
    [2, 0],
    [0, 0],
    [0, 2]
])

# Labels: +1 or -1
y = np.array([1, 1, -1, -1])
```

```python
from cvxopt import matrix, solvers

# Compute Gram matrix (kernel for linear SVM)
m, n = X.shape
K = np.dot(X, X.T)
P_numpy = np.outer(y,y)*K
P = matrix(P_numpy,tc='d')
q = matrix(-np.ones(m))
G = matrix(-np.eye(m))
h = matrix(np.zeros(m))
A = matrix(y, (1, m), 'd')
b = matrix(0.0)

# Solve QP
sol = solvers.qp(P, q, G, h, A, b)
alphas = np.ravel(sol['x'])

# Support vectors have non-zero alpha
sv = alphas > 1e-5
ind = np.arange(len(alphas))[sv]
alpha_sv = alphas[sv]
X_sv = X[sv]
y_sv = y[sv]

# Compute weight vector
w = np.sum(alpha_sv[:, None] * y_sv[:, None] * X_sv, axis=0)

# Compute bias
b = np.mean([y_sv[i] - np.dot(w, X_sv[i]) for i in range(len(alpha_sv))])

print("Weight vector:", w)
print("Bias:", b)
```

```python
x_n=np.array([
    [2, 2],
    [0, 0],
    [0, 0],
    [0, 0]
])

y=np.sign(np.dot(x_n,w.T) +b)
y=['Positive' for x in y if x==+1 else 'negative']
```
