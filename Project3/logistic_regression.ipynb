### Lets work on some churn datasets

```python
df_churn=pd.read_csv('Datasets/churn-bigml-20.csv')
df_churn.head()
```

```python
df_churn.columns
```

```python
df_churn.dtypes
```

```python
df_churn.info()
```

```python
df_churn['Voice mail plan']=df_churn['Voice mail plan'].apply(lambda x:1 if x=='Yes'
                                                              else 0)
```

```python
df_churn['International plan']=df_churn['International plan'].apply(lambda x:1 if x=='Yes'
                                                              else 0)
```

```python
df_churn.info()
```

```python
#select useful features
from sklearn.feature_selection import SelectKBest, chi2
```

```python
dummies=pd.get_dummies(df_churn.State,dtype=int)
dummies.head()
df_churn2=pd.concat([df_churn,dummies],axis=1)
```

```python
df_churn2.drop(columns=['State','LA'],inplace=True)
```

```python
X=df_churn2.drop('Churn',axis=1)
y=df_churn2['Churn']
```

```python
chi2_selector = SelectKBest(score_func=chi2, k=5)
X_kbest = chi2_selector.fit_transform(X, y)
```

```python
selected_features = X.columns[chi2_selector.get_support()]
selected_features
```

```python
df_churn2.shape
```

```python
df_churn3=df_churn2[selected_features]
```

```python
for item in selected_features:
    plt.figure()  # Create a new figure
    plt.scatter(x=df_churn3[item], y=y, marker='+', color='red')
    plt.title(f'Scatter plot of {item} vs y')
    plt.xlabel(item)
    plt.ylabel('y')
    plt.show()
```

```python
X_train, X_test, y_train, y_test=train_test_split(df_churn3,y,test_size=0.2,train_size=0.8,random_state=4)
```

```python
X_train
```

```python
y_train
```

```python
model2=linear_model.LogisticRegression(class_weight='balanced')
```

```python
model2.fit(X_train,y_train)
```

```python
y_pred=model2.predict(X_test)
```

```python
model2.score(X_test,y_test)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cm = confusion_matrix(y_test, y_pred)

# Display it
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap='Reds')
plt.title('Confusion Matrix')
plt.show()
```
