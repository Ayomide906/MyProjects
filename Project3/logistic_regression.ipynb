```python
import pandas as pd
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
df= pd.read_csv('Datasets/insurance.csv')
df.head()
```

```python
plt.scatter(data=df,x='age',y='bought_insurance',marker='+',color='red')
plt.show()
```

```python
df.shape
```

```python
#split test train
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(df[['age']],df.bought_insurance,test_size=0.1,train_size=0.9)
```

```python
y_train
```

```python
X_test
```

```python
from sklearn import linear_model
model=linear_model.LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.predict(X_test)
```

```python
model.score(X_test,y_test)
```

```python
model.predict_proba(X_test)
```

```python
df_churn=pd.read_csv('Datasets/churn-bigml-20.csv')
df_churn.head()
```

```python
df_churn.columns
```

```python
df_churn.dtypes
```

```python
df_churn.info()
```

```python
df_churn['Voice mail plan']=df_churn['Voice mail plan'].apply(lambda x:1 if x=='Yes'
                                                              else 0)
```

```python
df_churn['International plan']=df_churn['International plan'].apply(lambda x:1 if x=='Yes'
                                                              else 0)
```

```python
df_churn.info()
```

```python
#select useful features
from sklearn.feature_selection import SelectKBest, chi2
```

```python
dummies=pd.get_dummies(df_churn.State,dtype=int)
dummies.head()
df_churn2=pd.concat([df_churn,dummies],axis=1)
```

```python
df_churn2.drop(columns=['State','LA'],inplace=True)
```

```python
X=df_churn2.drop('Churn',axis=1)
y=df_churn2['Churn']
```

```python
chi2_selector = SelectKBest(score_func=chi2, k=5)
X_kbest = chi2_selector.fit_transform(X, y)
```

```python
selected_features = X.columns[chi2_selector.get_support()]
selected_features
```
