```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn import svm
```

```python
iris= load_iris()
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df['flower']=iris.target
df['flower_name']=df['flower'].apply(lambda x : iris.target_names[x])
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test= train_test_split(iris.data, iris.target,test_size=0.3)
```

```python
model=svm.SVC(C=20,gamma='auto')
model.fit(X_train,y_train)
model.score(X_test,y_test)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
cross_val_score(svm.SVC(kernel='linear',C=10,gamma='auto'),iris.data,iris.target,cv=5)
```

```python
import numpy as np
```

```python
arr=cross_val_score(svm.SVC(kernel='rbf',C=20,gamma='auto'),iris.data,iris.target,cv=5)
print(np.average(arr))
```

```python
kernels=['rbf','linear']
Cs=[1,10,20,30]
avg={}
for k in kernels:
    for C in Cs:
        cv_score=cross_val_score(svm.SVC(kernel=k,C=C,gamma='auto'),iris.data,iris.target,cv=5)
        avg[k + '_' + str(C)]=np.average(cv_score)
avg
        
```
