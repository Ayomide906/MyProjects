## Linear Regression with Multiple Variables

```python
df_mv=pd.read_csv('Datasets/homeprices_multivar.csv')
df_mv.head()
```

```python
import math
df_mv.bedrooms.fillna(math.floor(df_mv.bedrooms.median()),inplace=True)
```

```python
df_mv.head()
```

```python
df_mv.corr(numeric_only=True)
```

```python
import seaborn as sns
sns.pairplot(df_mv)
plt.show()
```

```python
model=linear_model.LinearRegression()
model.fit(df_mv[['area','bedrooms','age']],df_mv['price'])
```

```python
model.coef_
```

```python
model.intercept_
```

```python
model.predict([[3000,3,40],[2500,6,30]])
```

```python
model.predict([[2600,3,20]])
```

Underestimated the price but the difference isnt so large

```python
import pickle
```

```python
#to dump a model
with open('model_pickle','wb') as f: #wb mean write mode of a binary file
    pickle.dump(model,f)
```

```python
#to load a model
with open('model_pickle','rb') as f:
    np=pickle.load(f)
```

```python
np.predict([[3000,3,40],[2500,6,30]])
```

```python
#using joblib
import joblib
joblib.dump(model,'regression_model.pkl')
```

```python
npa=joblib.load('regression_model.pkl')
```

```python
npa.predict([[3000,3,40],[2500,6,30]]).round(2)
```
