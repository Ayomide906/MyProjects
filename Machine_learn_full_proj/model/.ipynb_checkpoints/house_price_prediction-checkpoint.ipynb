```python
def remove_outliers(df,col):
    Q1=df[col].quantile(0.25)
    Q3=df[col].quantile(0.75)
    iqr=Q3-Q1
    df_no_outlier=df[(df[col]>Q1-(1.5*iqr))&
                     (df[col]<Q3+(1.5*iqr))]
    return df_no_outlier
    
```

```python
for col in df_num.columns:
    df_new=remove_outliers(df_num,col)
df_new.info()
```

```python
def is_float(x):
    try:
        float(x)
    except:
        return False
    return True
```

```python
import re

def convert_to_sqft(x):
    x = str(x).strip()

    # Case 1: Sq. Yard → sq ft
    if 'Sq. Yard' in x or 'Sq. Yards' in x:
        num = float(re.sub('[^0-9.]', '', x).rstrip('.'))
        return num * 9

    # Case 2: Sq. Meter → sq ft
    if 'Sq. Meter' in x or 'Sq. Metre' in x:
        num = float(re.sub('[^0-9.]', '', x).rstrip('.'))
        return num * 10.7639

    # Case 3: range like "2100-2850"
    if '-' in x:
        cleaned = re.sub('[^0-9.\-]', '', x)
        low, high = cleaned.split('-')
        return (float(low.rstrip('.')) + float(high.rstrip('.'))) / 2

    # Case 4: plain number or number with trailing dot
    cleaned = re.sub('[^0-9.]', '', x).rstrip('.')
    try:
        return float(cleaned)
    except:
        return None
```

## Feature Engineering
### 1. Price per sq ft

```python
df3=df2.copy()
```

```python
df3['price_per_sqft']=df3.apply(lambda x: x['price']*100000/x['total_sqft'], axis=1)
df3.head()
```

```python
df3['location']=df3['location'].apply(lambda x:x.strip())
```

```python
location_stat=df3.groupby('location')['location'].count().sort_values(ascending=False)
location_stat_less=location_stat[location_stat<9]
location_stat_less
df3['location']=df3['location'].apply(lambda x: 'other' if x in location_stat_less
                                     else x)
```

```python
df3.sample(10)
```

### More Outlier removal 

```python
df4=df3[~(df3['total_sqft']/df3.bhk<300)]
df4.shape
```

```python
for key, df in df4.groupby('location'):
    print(key)
    print('/n')
    print(df)
```

```python
def remove_pps_outliers(df):
    df_out=pd.DataFrame()
    for key, subdf in df.groupby('location'):
        mean=subdf.price_per_sqft.mean()
        std=subdf['price_per_sqft'].std()
        red_df=subdf[(df['price_per_sqft']>mean-2*std)&(df['price_per_sqft']<mean+2*std)]
        df_out=pd.concat([df_out,red_df],ignore_index=True)
    return df_out
    
```

```python
df5=remove_pps_outliers(df4)
```

```python
df5.groupby('location').count()
```

```python
def plot_scatter(df,location):
    bhk2=df[(df.location==location)&(df.bhk==2)]
    bhk3=df[(df.location==location)&(df.bhk==3)]
    plt.figure(figsize=(10,8))
    plt.scatter(x=bhk2.total_sqft,y=bhk2.price,color='blue',label='2 bhk')
    plt.scatter(x=bhk3.total_sqft,y=bhk3.price,color='green',label='3 bhk',marker='+')
    plt.xlabel('total sqr_ft')
    plt.ylabel('price')
    plt.legend()
    plt.show()
plot_scatter(df5,'Yelahanka')
```

```python
df5.shape
```

```python
mean_bhk = df5.groupby(['location','bhk']).price_per_sqft.mean()
def remove_bhk_outliers(df):
    df_out = pd.DataFrame()
    for (loc, subdf) in df.groupby('location'):
        for bhk, bhk_df in subdf.groupby('bhk'):
            if bhk - 1 in subdf['bhk'].unique():
                prev_mean = mean_bhk[(loc, bhk - 1)]
                filtered = bhk_df[bhk_df['price_per_sqft'] > prev_mean]
                df_out = pd.concat([df_out, filtered], ignore_index=True)
            else:
                # No previous BHK exists i.e for 1 bhk
                df_out = pd.concat([df_out, bhk_df], ignore_index=True)
    
    return df_out

df6 = remove_bhk_outliers(df5)
df6.shape
```

### Train ML Model

```python
dum1=pd.get_dummies(df6.location,dtype=int)
```

```python
dum2=pd.get_dummies(df6.area_type,dtype=int)
```

```python
dum=pd.concat([dum1,dum2],axis=1)
dum.head()
```

```python
df7=pd.concat([df6,dum],axis=1)
df7.drop(columns=['location','area_type','price_per_sqft'],inplace=True)
df7.head()
```

```python
from sklearn.preprocessing import StandardScaler
X=df7.drop('price',axis='columns')
scaler= StandardScaler()
X_scaled=scaler.fit_transform(X)
X_scaled[:3]
```

```python
y=df7.price
y.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=10)
```

```python
from sklearn.linear_model import LinearRegression
lr_clf=LinearRegression()
lr_clf.fit(X_train,y_train)
lr_clf.score(X_test,y_test)
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import ShuffleSplit
cv = ShuffleSplit(n_splits=5,test_size=0.2,random_state=0)
cross_val_score(LinearRegression(),X,y,cv=cv)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Lasso
from sklearn.ensemble import BaggingRegressor
from sklearn.tree import DecisionTreeRegressor
```

```python
model_params={
    'Lasso':{
        'model':Lasso(),
        'params':{
            'alpha':[i for i in range(1,6)]
        }
    },
    'DTR':{
        'model':DecisionTreeRegressor(),
        'params':{
            'criterion':['squared_error','friedman_mse','absolute_error','poisson']
        }
    }
}
```

```python
scores=[]
for model_name,np in model_params.items():
    clf=GridSearchCV(estimator=np['model'],param_grid=np['params'],cv=5)
    clf.fit(X_scaled,y)
    scores.append({
        'model':model_name,
        'best_score':clf.best_score_,
        'best_param':clf.best_params_
    })
    
```

```python
pd.DataFrame(scores,columns=['model','best_score','best_params'])
```

```python
bag_model=BaggingRegressor(
    estimator=LinearRegression(),
    n_estimators=50,
    max_samples=0.8,
    oob_score=True
)
```

```python
bag_model.fit(X_train,y_train)
bag_model.oob_score_
```

```python
bag_model.score(X_test,y_test)
```

```python
from sklearn.pipeline import Pipeline
```

```python
Pi_model=Pipeline(
    [
        ('scaler',StandardScaler()),
        ('model',LinearRegression())
    ]
)
```

```python
Pi_model.fit(X_train,y_train)
Pi_model.score(X_test,y_test)
```

```python
def predict_price(loc, sqft, bath, bhk):
    # Create zero vector for all columns
    x = np.zeros(len(X.columns))

    # Fill numeric features
    x[0] = bath
    x[1] = sqft
    x[2] = bhk

    # Check if the location exists in the columns
    if loc in X.columns:
        loc_index = np.where(X.columns == loc)[0][0]
        x[loc_index] = 1
    else:
        if "others" in X.columns:
            other_index = np.where(X.columns == "others")[0][0]
            x[other_index] = 1

    return Pi_model.predict([x])[0]
```

```python
predict_price('1st Phase JP Nagar',1500,2,1)
```
